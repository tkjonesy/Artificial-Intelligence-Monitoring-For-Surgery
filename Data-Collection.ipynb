```python
import cv2 as cv
import time
```

```python
# Initialize the video capture
cap = cv.VideoCapture(0)
```

```python
# MAKE SURE THERE IS A '/' AT THE END OF THE FILE PATH
IMAGESPERBATCH = 600 # How many images to be taken per run
IMAGESBEFOREPAUSE = 25 # How many images to be taken before pausing
IMAGEINDEXOFFSET = 0 # Adds an offset to image name (if there are 400 images already named, offset should be set to 400)
PATH = 'C:/Users/Trever/Desktop/Dataset/Final/Raw/Background/'  # Location of images to be saved

# Main loop for taking images
x = 0
while x < IMAGESPERBATCH:
    # Pause every i images
    if x == 0 or x % IMAGESBEFOREPAUSE == 0:
        while True:
            # Capture each frame from the webcam
            ret, image = cap.read()
            
            # Display the frame
            cv.imshow('Image Capture', image)
        
            # Break on pressing Q key
            if cv.waitKey(1) & 0xFF == ord('q'):
                break
        
        cv.destroyAllWindows()
    
    # Capture each frame from the webcam
    ret, image = cap.read()
    
    cv.imwrite(f'{PATH}{x + IMAGEINDEXOFFSET + 1}.jpg', image)
    
    x += 1
    time.sleep(.85)

# Release resources
cap.release()
```
