```python
import cv2 as cv
import os
```

```python
TARGET = 'C:/Users/Trever/Desktop/Dataset/Raw/Multiclass/'  # Location of images that need to be resized
DEST = 'C:/Users/Trever/Desktop/Dataset/Resized/Multiclass/'  # Location of resized images to be saved

fileNames = [f for f in os.listdir(TARGET)]

for file in fileNames:
    path = TARGET + file
    img = cv.imread(path)

    # Get original dimensions
    h, w = img.shape[:2]

    # Compute new width and height while keeping aspect ratio
    scale = min(640 / w, 640 / h)
    new_w = int(w * scale)
    new_h = int(h * scale)

    resized = cv.resize(img, (new_w, new_h))

    # Create a new 640x640 image with padding
    result = cv.copyMakeBorder(
        resized,
        top=(640 - new_h) // 2,
        bottom=(640 - new_h + 1) // 2,
        left=(640 - new_w) // 2,
        right=(640 - new_w + 1) // 2,
        borderType=cv.BORDER_CONSTANT,
        value=(0, 0, 0)  # Black padding
    )

    cv.imwrite(DEST + file, result)
```
